```python
x = float(input())

phan_nguyen = int(x)
phan_thap_phan = x - phan_nguyen

if phan_thap_phan == 0:
    tron_xuong = phan_nguyen
    tron_len = phan_nguyen
    gan_nhat = phan_nguyen
elif x > 0:
    tron_xuong = phan_nguyen
    tron_len = phan_nguyen + 1
    gan_nhat = phan_nguyen + 1 if phan_thap_phan >= 0.5 else phan_nguyen
else:
    tron_len = phan_nguyen
    tron_xuong = phan_nguyen - 1
    gan_nhat = phan_nguyen - 1 if phan_thap_phan <= -0.5 else phan_nguyen

print(tron_len, tron_xuong, gan_nhat)
```
