```python
import sys

for line in sys.stdin:
    parts = line.split()
    if len(parts) < 2:
        continue
    a = int(parts[0])
    b = int(parts[1])
    tong = 0
    for x in range(a, b + 1):
        if x >= 2:
            nt = True
            i = 2
            while i * i <= x:
                if x % i == 0:
                    nt = False
                    break
                i += 1
            if nt:
                tong += x
    print(tong)
```

```python
n = int(input())
buoc = 0

while True:
    temp = n
    rev = 0
    while temp > 0:
        rev = rev * 10 + temp % 10
        temp //= 10
    if n == rev:
        break
    n = n + rev
    buoc += 1

print(buoc, n)
```

```python
n = int(input())
co_so = False
i = 1
while i * i <= n:
    cp = i * i
    s = str(cp)
    khac_nhau = True
    for j in range(len(s)):
        for k in range(j + 1, len(s)):
            if s[j] == s[k]:
                khac_nhau = False
                break
        if not khac_nhau:
            break
    if khac_nhau:
        print(cp, end=" ")
        co_so = True
    i += 1

if not co_so:
    print("(no number)")
```

```python
n = int(input())
best_x = 1
max_len = 1

for x in range(1, n + 1):
    curr = x
    length = 1
    while curr != 1:
        if curr % 2 == 0:
            curr //= 2
        else:
            curr = 3 * curr + 1
        length += 1
    if length > max_len:
        max_len = length
        best_x = x

print(best_x, max_len)
```

```python
n = int(input())
dem = 0
i = 1
while i * i <= n:
    if n % i == 0:
        if i % 2 == 0:
            dem += 1
        uoc_kia = n // i
        if uoc_kia != i and uoc_kia % 2 == 0:
            dem += 1
    i += 1
print(dem)
```

```python
a = int(input())
b = int(input())

tong_a = 0
for i in range(1, a):
    if a % i == 0:
        tong_a += i

tong_b = 0
for i in range(1, b):
    if b % i == 0:
        tong_b += i

if tong_a == b and tong_b == a:
    print("true")
else:
    print("false")
```
