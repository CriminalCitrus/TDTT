```python
h = int(input())
for i in range(1, h + 1):
    if i == 1:
        print(" " * (h - 1) + "*")
    elif i == h:
        print("*" * (2 * h - 1))
    else:
        print(" " * (h - i) + "*" + " " * (2 * i - 3) + "*")
```

```python
first = True
max_val = 0
min_val = 0

while True:
    x = int(input())
    if x == -1:
        break
    if first:
        max_val = x
        min_val = x
        first = False
    else:
        if x > max_val:
            max_val = x
        if x < min_val:
            min_val = x

if not first:
    print(max_val, min_val)
```

```python
n = int(input())

# Cach 1: Chia nguyen cho 10
temp = n
dem1 = 0
while temp > 0:
    dem1 += 1
    temp //= 10
print(dem1)

# Cach 2: So sanh voi luy thua cua 10
dem2 = 1
moc = 10
while moc <= n:
    dem2 += 1
    moc *= 10
print(dem2)

# Cach 3: Chuyen sang chuoi (str)
dem3 = 0
for ch in str(n):
    dem3 += 1
print(dem3)
```

```python
n = int(input())
chan = 0
le = 0
while n > 0:
    chu_so = n % 10
    if chu_so % 2 == 0:
        chan += 1
    else:
        le += 1
    n //= 10
print(chan, le)
```
